```python
from sympy import *
import numpy as np
import matplotlib.pyplot as plt
```

```python
t = symbols('t')
x = Function('x')
```

```python
s = dsolve(250*x(t).diff(t, 2) + 12*x(t).diff(t) + 24, x(t),
           ics={x(0): 0, x(t).diff(t).subs(t, 0): 8})
```

```python
xt = s.rhs
vt = xt.diff(t)
T_return = nsolve(xt, t, 20)
print(f"Ответ: лодка вернется к пирсу через t = {T_return:.4f} секунд")

print("x(t) =", xt)
print("v(t) =", vt)
```

```python
T = float(T_return)
fx = lambdify(t, xt, 'numpy')
fv = lambdify(t, vt, 'numpy')
tt = np.linspace(0, T, 300)
```

```python
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.plot(tt, fx(tt), label='x(t), м')
plt.axhline(0, color='black', linestyle='--')
plt.title('Координата лодки')
plt.xlabel('t, с')
plt.grid()
```

```python
plt.subplot(1, 2, 2)
plt.plot(tt, fv(tt), label='v(t), м/с', color='red')
plt.axhline(0, color='black', linestyle='--')
plt.title('Скорость лодки')
plt.xlabel('t, с')
plt.grid()
```

```python
plt.tight_layout()
plt.show()
```
